```python
# Notebook: nb_bronze_jse_sarb_load
# Purpose: Ingest JSE and SARB raw CSV files into Bronze Delta tables

print("Starting JSE and SARB Bronze Ingestion...")
```

```python
# 1. Ingest JSE Listings 
print("\n Processing JSE Listings...")
jse_file_path = "Files/bronze/jse_listings/jse_listings_sample.csv"
```

```python
# Read the raw CSV from the Lakehouse Files section
jse_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(jse_file_path)

# Write as a managed Delta table in the Tables section
# mode("overwrite") ensures we can re-run this notebook without errors
jse_df.write.format("delta").mode("overwrite").saveAsTable("bronze_jse_listings")
print(f" Successfully created 'bronze_jse_listings' with {jse_df.count()} rows.")
```

```python
# 2. Ingest SARB Bank Stats 
print("\n Processing SARB Bank Stats...")
sarb_file_path = "Files/bronze/sarb_stats/sarb_bank_stats_sample.csv"
```

```python
# Read the raw CSV
sarb_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(sarb_file_path)

# Write as a managed Delta table
sarb_df.write.format("delta").mode("overwrite").saveAsTable("bronze_sarb_stats")
print(f" Successfully created 'bronze_sarb_stats' with {sarb_df.count()} rows.")

print("\n Bronze Ingestion Complete!")
```
